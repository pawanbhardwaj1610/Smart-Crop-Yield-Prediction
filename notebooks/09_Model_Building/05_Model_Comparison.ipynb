```python
import pandas as pd

comparison = pd.read_csv(
    "../models/model_comparison.csv"
)

comparison
```

# Method 1: Select by Highest R² Score (Most Common)


```python
best_model = comparison.loc[
    comparison["R2_Score"].idxmax()
]

best_model
```

# Method 2: Select by Lowest Error

```python
best_mae_model = comparison.loc[
    comparison["MAE"].idxmin()
]

best_mae_model
```

```python
best_rmse_model = comparison.loc[
    comparison["RMSE"].idxmin()
]

best_rmse_model
```

```python
comparison["Score"] = (
    comparison["R2_Score"]
    - comparison["MAE"]*0.01
    - comparison["RMSE"]*0.01
)

comparison
```

```python
best_model = comparison.loc[
    comparison["Score"].idxmax()
]

best_model
```
